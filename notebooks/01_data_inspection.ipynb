### 16. Cleaning dataset

#### 1. Cleaning tanggal

```python
import pandas as pd

def clean_transaction_date(date_value):

    if pd.isna(date_value):
        return pd.NaT

    date_value = str(date_value).strip()

    formats = [
        "%Y-%m-%d",
        "%d/%m/%Y",
        "%m-%d-%Y",
        "%d-%b-%Y"
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(
                date_value,
                format=fmt
            )
        except ValueError:
            continue

    return pd.NaT
```

```python
df["transaction_date_clean"] = (
    df["transaction_date"]
    .apply(clean_transaction_date)
)

print(df[
    ["transaction_date", "transaction_date_clean"]
].head(30))
```

## Data Cleaning

#### 1. Dataset Cleaning

```python
# Membuat salinan dataset
df_clean = df.copy()

print("Jumlah baris sebelum cleaning:", len(df_clean))
```

#### 2. Normalisasi nama merek dan body style

```python
# Standarisasi nama merek
df_clean["make"] = (
    df_clean["make"]
    .str.strip()
    .str.lower()
)

# Standarisasi body style
df_clean["body-style"] = (
    df_clean["body-style"]
    .str.strip()
    .str.lower()
)

print("Kategori make setelah cleaning:")
print(sorted(df_clean["make"].dropna().unique()))

print("\nKategori body-style:")
print(df_clean["body-style"].value_counts(dropna=False))
```

#### 3. Menghapus data duplicate

```python
before = len(df_clean)

df_clean = df_clean.drop_duplicates(keep="first") 

after = len(df_clean)

print("Baris sebelum:", before)
print("Baris sesudah:", after)
print("Duplikat dihapus:", before - after)
print("Sisa duplikat:", df_clean.duplicated().sum())
```

#### 4. Imputasi

```python
# Periksa modus jumlah pintu berdasarkan body style
door_mode_by_body = (
    df_clean.groupby("body-style")["num-of-doors"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
)

print(door_mode_by_body)
```

```python
missing_doors = df_clean["num-of-doors"].isnull()

df_clean.loc[missing_doors, "num-of-doors"] = (
    df_clean.loc[missing_doors, "body-style"]
    .map(door_mode_by_body)
)

print("Missing num-of-doors setelah cleaning:",
      df_clean["num-of-doors"].isnull().sum())
```

#### 5. Missing value numerik


```python
# Periksa missing value setelah cleaning awal

print("Missing value sebelum imputasi:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# Mengisi missing value numerik menggunakan median

numeric_columns = ["stroke", "horsepower", "price"]

for column in numeric_columns:
    median_value = df_clean[column].median()

    df_clean[column] = df_clean[column].fillna(median_value)

    print(f"{column}: missing value diisi dengan median {median_value:.2f}")
```

```python
print("Missing value setelah imputasi:")

print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

#### 6. Merapikan kolom make

```python
# Mengisi make yang kosong dengan unknown

df_clean["make"] = df_clean["make"].fillna("unknown")

# Menghapus kolom bantuan yang tidak diperlukan
df_clean = df_clean.drop(
    columns=["make_normalized"],
    errors="ignore"
)

print("Missing make:", df_clean["make"].isnull().sum())
print("Kategori make:")
print(sorted(df_clean["make"].unique()))
```

```python
print("Distribusi horsepower:")
print(df_clean["horsepower"].describe())

print("\nKategori horsepower-binned:")
print(df_clean["horsepower-binned"].value_counts(dropna=False))

print("\nContoh hubungan horsepower dan bin:")
print(
    df_clean[["horsepower", "horsepower-binned"]]
    .sort_values("horsepower")
    .head(20)
)
```

```python
print("Tipe data tanggal:")
print(df_clean["transaction_date_clean"].dtype)

print("\nJumlah tanggal kosong:")
print(df_clean["transaction_date_clean"].isnull().sum())

print("\nContoh tanggal:")
print(
    df_clean[["transaction_date", "transaction_date_clean"]]
    .head(15)
)
```

```python
# Melihat rentang horsepower pada setiap kategori

bin_summary = (
    df_clean.groupby("horsepower-binned", observed=False)["horsepower"]
    .agg(["min", "max", "count"])
)

print(bin_summary)
```

```python
print(
    df_clean[df_clean["horsepower-binned"].isnull()]
    [["horsepower", "horsepower-binned"]]
)
```

```python
# Membuat ulang kategori horsepower-binned

bins = [0, 100, 150, float("inf")]
labels = ["Low", "Medium", "High"]

df_clean["horsepower-binned"] = pd.cut(
    df_clean["horsepower"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Periksa hasil
print(df_clean["horsepower-binned"].value_counts())
print("\nMissing horsepower-binned:")
print(df_clean["horsepower-binned"].isnull().sum())
```
